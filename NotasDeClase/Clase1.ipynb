# Práctica


Investigar tipos de nubes y en qué se diferencian, cómo se ven, cómo reconocerlas desde la tierra

 Referencia : [Cloud Types](https://scied.ucar.edu/learning-zone/clouds/cloud-types)

Clouds are given different names based on their **shape** and their **height** in the sky. 

Most clouds can be divided into groups **(high/middle/low)** based on the height of the cloud's base above the Earth's surface.

[NOAA Ten Basic Clouds](https://www.noaa.gov/jetstream/clouds/ten-basic-clouds)

### Nubes Altas (Compuestas mayormente por cristales de hielo debido a su altura)
*   **Cirrus:** Se identifican como hebras blancas, finas y plumosas que parecen mechones de cabello.
*   **Cirrocumulus:** Aparecen como pequeños parches o granos blancos agrupados en el cielo.
*   **Cirrostratus:** Forman una capa delgada, blanquecina y transparente que suele cubrir gran parte del cielo.

### Nubes Medias (Compuestas por gotas de agua y algunos cristales de hielo)
*   **Altocumulus:** Se ven como masas redondas o parches blancos y grisáceos, más grandes que los cirrocúmulos.
*   **Altostratus:** Es una capa uniforme de color gris o azulado que cubre el cielo, a través de la cual el sol puede verse opaco.

### Nubes Bajas (Compuestas principalmente por gotas de agua)
*   **Stratus:** Se identifican como una capa grisácea y uniforme que parece neblina que no toca el suelo.
*   **Stratocumulus:** Son capas de nubes bajas que presentan masas redondeadas u oscuras.
*   **Cumulus:** Son las clásicas nubes blancas y esponjosas con base plana, similares a motas de algodón.
*   **Nimbostratus:** Nubes oscuras y densas de gran espesor que bloquean el sol y producen lluvia continua.

### Nubes de Desarrollo Vertical
*   **Cumulonimbus:** Nubes gigantescas en forma de torre o yunque; son las principales generadoras de tormentas y lluvias intensas.

## Composición

*   **Nubes Altas (Cirrus, Cirrocumulus y Cirrostratus):** Están compuestas casi exclusivamente por **cristales de hielo**.
*   **Nubes Medias (Altocumulus, Altostratus y Nimbostratus):** Se componen principalmente de **gotas de agua**, aunque pueden contener **cristales de hielo** si las temperaturas son lo suficientemente bajas.
*   **Nubes Bajas (Cumulus, Stratocumulus, Stratus y Cumulonimbus):** Están compuestas mayoritariamente por **gotas de agua**. En el caso de las **Cumulonimbus**, su gran desarrollo vertical permite que su parte superior también contenga hielo.

Para identificar los diez tipos básicos de nubes desde la tierra, puedes observar su forma, su efecto sobre la luz solar y su tamaño aparente utilizando tus dedos como referencia:

### Nubes Altas (Casi siempre blancas y transparentes)
*   **Cirrus (Ci):** Se ven como **filamentos blancos y delicados**, parches o bandas estrechas con una apariencia fibrosa (como cabello) o un brillo sedoso. No suelen disminuir el brillo del sol cuando pasan frente a él.
*   **Cirrocumulus (Cc):** Aparecen como una capa delgada de **granos o rizos blancos** dispuestos regularmente sin sombras. Sus elementos son muy pequeños, con un ancho menor al de tu **dedo meñique** con el brazo extendido.
*   **Cirrostratus (Cs):** Es un velo blanquecino y transparente que puede cubrir todo el cielo. Se identifica fácilmente porque casi siempre produce un **fenómeno de halo** (un anillo de luz) alrededor del sol o la luna.

### Nubes Medias (Grisáceas o blancas)
*   **Altocumulus (Ac):** Capas o parches de **masas redondeadas o rollos** blancos y grises. Sus elementos tienen un ancho de entre uno y cinco grados (entre el grosor de **un dedo y tres dedos** extendidos). Pueden formar una "corona" (anillo coloreado) muy cerca del sol.
*   **Altostratus (As):** Una capa uniforme gris o azulada que cubre total o parcialmente el cielo. El sol se ve a través de ella de forma opaca, como si fuera visto a través de un **vidrio esmerilado**, y no produce halos ni sombras en el suelo.
*   **Nimbostratus (Ns):** Es una capa de nube **gris oscura y densa** que bloquea completamente el sol. Se identifica por estar acompañada de **lluvia o nieve continua**, lo que le da una apariencia difusa.

### Nubes Bajas (Grisáceas o blancas brillantes)
*   **Cumulus (Cu):** Nubes individuales y densas con contornos definidos. Crecen verticalmente en forma de **cúpulas o torres**, y su parte superior suele parecerse a una **coliflor** blanca brillante mientras que su base es oscura y plana.
*   **Stratocumulus (Sc):** Parches o capas de nubes grises o blanquecinas que muestran **masas redondeadas u oscuras** en forma de panal. Sus elementos son grandes, con un ancho mayor a **tres dedos** extendidos.
*   **Stratus (St):** Una capa gris muy uniforme y plana que parece **niebla que no toca el suelo**. Si el sol es visible a través de ella, su contorno se distingue claramente.
*   **Cumulonimbus (Cb):** Es la nube de tormenta, una **torre gigante y densa** con una base muy oscura. Su parte superior suele ser plana y extenderse hacia los lados en forma de **yunque**.
