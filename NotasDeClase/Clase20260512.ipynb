Guía de refracción atmosférica.

### 1. Fundamentación Teórica según Karttunen

La refracción es el fenómeno de cambio en la dirección de un rayo de luz al pasar de un medio a otro con diferente velocidad de propagación. En el contexto astronómico, la luz viaja desde el vacío del espacio ($n=1$) e ingresa a la atmósfera terrestre, donde el índice de refracción aumenta progresivamente a medida que el aire se vuelve más denso cerca de la superficie.

#### El modelo de capas planas
Para objetos que no están demasiado cerca del horizonte, Karttunen explica que la atmósfera puede aproximarse como una **pila de capas planas paralelas**, cada una con su propio índice de refracción ($n_i$). Al aplicar la **ley de Snell** en cada una de estas capas, se obtiene que el producto del índice de refracción por el seno del ángulo respecto a la vertical es constante:
$$n_k \sin z_k = \dots = n_0 \sin \zeta = \sin z$$
Donde:
*   **$z$**: Distancia cenital real (fuera de la atmósfera).
*   **$\zeta$**: Distancia cenital aparente (observada desde el suelo).
*   **$n_0$**: Índice de refracción del aire a nivel del observador.

#### El ángulo de refracción ($R$)
El ángulo de refracción se define como la diferencia entre la distancia cenital real y la aparente ($R = z - \zeta$). Para ángulos pequeños, Karttunen deriva una aproximación fundamental:
$$R \approx (n_0 - 1) \tan \zeta$$
Este valor de $R$ se expresa habitualmente en radianes en las fórmulas teóricas, aunque en la práctica se convierte a minutos o segundos de arco.

### 2. Factores que afectan la refracción

La magnitud de la refracción no es constante; depende críticamente de las condiciones físicas del aire a lo largo de la línea de visión.

*   **Presión y Temperatura:** El índice de refracción depende de la densidad del aire. Karttunen proporciona una fórmula práctica para altitudes superiores a $15^\circ$:
    $$R \approx \frac{P}{273 + T} \times 0.00452^\circ \tan(90^\circ - a)$$
    Donde $P$ es la presión en hectopascales, $T$ es la temperatura en grados Celsius y $a$ es la altura del objeto.
*   **Curvatura de la atmósfera:** A alturas muy bajas (cerca del horizonte), el modelo de capas planas falla y es necesario considerar la curvatura terrestre, lo que requiere fórmulas más complejas que tienen en cuenta el camino óptico más largo a través de la atmósfera.
*   **Cenit:** En el cenit (directamente sobre la cabeza), la refracción es nula ya que la luz entra perpendicularmente a las capas atmosféricas.

### 3. Aplicaciones y Manifestaciones en Astronomía

La refracción tiene implicaciones profundas en la forma en que medimos y percibimos el cosmos:

*   **Elevación aparente de los astros:** La refracción siempre hace que un objeto parezca estar **más alto** en el cielo de lo que realmente está. Este efecto es máximo en el horizonte, donde la refracción alcanza aproximadamente **34 minutos de arco**, valor que es ligeramente superior al diámetro aparente del Sol o la Luna.
*   **Cálculo de amaneceres y atardeceres:** Debido a la refracción horizontal de 34', vemos el Sol antes de que su disco cruce geométricamente el horizonte por la mañana, y lo seguimos viendo después de que ya se ha puesto por la tarde. Por esta razón, los almanaques astronómicos ajustan las horas de salida y puesta utilizando una altura de $-50'$ (considerando 34' de refracción y 16' de semidiámetro solar).
*   **Deformación de discos extensos:** Cuando el Sol o la Luna están cerca del horizonte, la parte inferior del disco está a una altura menor que la parte superior y, por lo tanto, sufre una refracción más fuerte. Esto resulta en un **achatamiento visual** del disco, dándole una apariencia ovalada.
*   **Reducción de observaciones:** Los catálogos estelares proporcionan "posiciones medias" (libres de efectos locales). El astrónomo debe aplicar correcciones por refracción a sus medidas en bruto para poder comparar sus datos con los catálogos fundamentales, o viceversa, para saber dónde apuntar un telescopio.
*   **Scintillation (Centelleo):** Karttunen vincula la turbulencia atmosférica con cambios rápidos en la refracción que causan que las estrellas parezcan cambiar de brillo y posición rápidamente (titileo). Este efecto es más pronunciado cerca del horizonte debido a la mayor cantidad de atmósfera atravesada.
*   **Seeing:** La atmósfera limita la resolución de los telescopios. Debido a la refracción irregular, la imagen de una estrella se esparce en un "disco de seeing" en lugar de ser un punto perfecto, lo que impide alcanzar el límite de difracción teórico del instrumento en la mayoría de los sitios a nivel del mar.
