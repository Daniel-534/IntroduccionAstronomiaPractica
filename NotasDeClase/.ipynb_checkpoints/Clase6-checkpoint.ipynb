Explicación de las razones por las cuales se ve la luna en cada una de sus fases.

* Cuarto menguante
* Luna nueva
* Cuarto creciente
* Llena

Posición relativa entre el sol y la luna

* Conjunción

Posición aparente del sol

Velocidad angular de la luna -> $\sim 13°/ \text{día}$ (360/27 // N)

Velocidad angular de la tierra -> $\sim 15°/ \text{h}$

Cada día la luna sale 52 minutos más tarde

Entre cada fase hay aproximadamente 7 días

Fecha de entrega de este informe: 19 de Marzo (9 días)


Idea de la práctica:

$\omega = 15°/\text{h}$
$\omega = \frac{\theta}{t} \longrightarrow \theta = \omega t$

```
AstroGraphics[
 {
  {Red, AstroStep[Sun]}, 
  {Blue, AstroStep[Earth]}, 
  {Gray, AstroStep[Moon]}
 }, 
 ViewPoint -> {1, 1, 1}, 
 Frame -> True, 
 Background -> Black
]
```

```
AstroGraphics[
 {
  AstroObject["Sun"],
  AstroObject["Earth"],
  AstroObject["Moon"]
 },
 PlotRange -> "SolarSystem",
 AstroCenter -> Earth,
 AstroProjection -> "Equatorial"
]
```

```
Rasterize[
 AstroGraphics[
  {AstroObject["Sun"], AstroObject["Earth"], AstroObject["Moon"]},
  PlotRange -> "SolarSystem",
  AstroCenter -> Earth,
  AstroProjection -> "Equatorial"
 ]
]
```
