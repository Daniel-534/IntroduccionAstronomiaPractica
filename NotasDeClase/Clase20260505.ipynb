### 1. Planeta Interior y Planeta Exterior
Karttunen clasifica los planetas en dos grupos principales dependiendo de la posición de sus órbitas en relación con la de la Tierra:
*   **Planeta Interior (Inferior Planet):** Son aquellos cuyas órbitas se encuentran dentro de la órbita terrestre. En nuestro sistema solar, los únicos planetas interiores son **Mercurio y Venus**. Estos planetas muestran un ciclo completo de fases (como la Luna) y siempre se observan en las cercanías del Sol, ya sea como "estrellas de la mañana" o "estrellas de la tarde".
*   **Planeta Exterior (Superior Planet):** Son los planetas cuyas órbitas se sitúan fuera de la órbita de la Tierra. Este grupo incluye a **Marte, Júpiter, Saturno, Urano y Neptuno**. A diferencia de los interiores, estos planetas no muestran un ciclo completo de fases visto desde la Tierra y pueden observarse en cualquier punto de la eclíptica, incluso a 180° del Sol.

### 2. Oposición y Conjunción
Estas son configuraciones geométricas que resultan del movimiento relativo de los planetas y la Tierra alrededor del Sol:
*   **Oposición:** Esta configuración es exclusiva de los **planetas superiores**. Ocurre cuando el planeta se encuentra exactamente en el lado opuesto al Sol visto desde la Tierra; es decir, la Tierra está posicionada directamente entre el Sol y el planeta. En este momento, la diferencia de longitudes eclípticas entre el planeta y el Sol es de 180°, lo que permite que el planeta sea visible durante toda la noche.
*   **Conjunción:** Se refiere a la alineación de un planeta con el Sol desde la perspectiva terrestre. 
    *   Para **planetas superiores**, hay una sola conjunción cuando el planeta se encuentra detrás del Sol. 
    *   Para **planetas interiores**, existen dos tipos: la **conjunción inferior**, cuando el planeta pasa entre la Tierra y el Sol, y la **conjunción superior**, cuando el planeta se encuentra detrás del Sol. En ambos casos de conjunción, las longitudes eclípticas del planeta y el Sol son iguales.

### 3. Periodo Sideral y Periodo Sinódico
Karttunen distingue entre el tiempo real de traslación de un planeta y el tiempo percibido entre eventos astronómicos sucesivos:
*   **Periodo Sideral ($P$):** Es el tiempo real que le toma a un planeta completar una revolución completa alrededor del Sol con respecto a las estrellas fijas. Se considera el periodo de revolución "verdadero" y es una propiedad intrínseca y única de la órbita de cada objeto celeste. Por ejemplo, el periodo sideral de la Tierra es de aproximadamente 365.25 días.
*   **Periodo Sinódico ($P_{1,2}$):** Es el intervalo de tiempo que transcurre entre dos configuraciones planetarias idénticas sucesivas vistas desde la Tierra, como el tiempo entre dos oposiciones o dos conjunciones consecutivas. Este periodo depende del movimiento relativo tanto de la Tierra como del planeta observado. Matemáticamente, se relaciona con los periodos siderales mediante la fórmula $1/P_{syn} = |1/P_1 - 1/P_2|$, donde $P_1$ y $P_2$ son los periodos siderales de la Tierra y el planeta.
